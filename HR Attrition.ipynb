# Model building 

```python
import numpy as np 
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score
```

```python
from sklearn.naive_bayes import GaussianNB, BernoulliNB
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC, LinearSVC
```

Encode categorical variables

```python
df_enc = employee_data.copy()
cat_cols = df_enc.select_dtypes(include=["object"]).columns.tolist()

encoders = {}
for col in cat_cols:
    le = LabelEncoder()
    df_enc[col] = le.fit_transform(df_enc[col])
    encoders[col] = dict(zip(le.classes_, le.transform(le.classes_)))
```

Features & Target

```python
X = df_enc.drop(columns=["Attrition"]).values
y = df_enc["Attrition"].values
```

Train/test split (70/30)

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, stratify=y, random_state=42
)
```

Standardize (Z-score scaling)

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

define model 

```python
models = {
    "Gaussian NB": GaussianNB(),
    "Bernoulli NB": BernoulliNB(),
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "KNN": KNeighborsClassifier(n_neighbors=5),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=200, random_state=42),
    "SVC": SVC(kernel="rbf", random_state=42),
    "Linear SVC": LinearSVC(max_iter=5000, random_state=42)
}
```

Train and evaluate 

```python
results = []

for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, zero_division=0)
    rec = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)
    cm = confusion_matrix(y_test, y_pred)

    results.append({
        "Model": name,
        "Accuracy": round(acc,3),
        "Precision": round(prec,3),
        "Recall": round(rec,3),
        "F1": round(f1,3),
        "ConfusionMatrix": cm
    })
```

Result

```python
res_df = pd.DataFrame(results).sort_values(by="Recall", ascending=False)
print("\n=== Model Performance (sorted by Recall) ===")
print(res_df[["Model","Accuracy","Precision","Recall","F1"]])

for r in results:
    print(f"\n{r['Model']} Confusion Matrix:\n{r['ConfusionMatrix']}")
```

# Model Evaluation 

```python
def compute_metrics(y_true, y_pred):
    cm = confusion_matrix(y_true, y_pred, labels=[0,1])
    tn, fp, fn, tp = cm.ravel()
    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, zero_division=0)
    recall = recall_score(y_true, y_pred, zero_division=0)
    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0.0   
    f1 = f1_score(y_true, y_pred, zero_division=0)
    return accuracy, precision, recall, specificity, f1, cm
```

```python
eval_results = []

for name, model in models.items():
    y_pred = model.predict(X_test)
    acc, prec, rec, spec, f1, cm = compute_metrics(y_test, y_pred)

    eval_results.append({
        "Model": name,
        "Accuracy": round(acc,3),
        "Precision": round(prec,3),
        "Recall": round(rec,3),
        "Specificity": round(spec,3),
        "F1 Score": round(f1,3),
        "Confusion Matrix": cm
    })
```

```python
results_df = pd.DataFrame(eval_results).sort_values(by="Recall", ascending=False)
print("Model Evaluation Results (sorted by Recall)")
print(results_df[["Model","Accuracy","Precision","Recall","Specificity","F1 Score"]])
```

```python
for r in eval_results:
    print(f"\n{r['Model']} Confusion Matrix (labels: 0=Stayed, 1=Left):\n{r['Confusion Matrix']}")
```

# Fine Tuning 

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC

param_grid = {
    'C': [0.1, 1, 10],
    'kernel': ['linear', 'rbf', 'poly'],
    'gamma': ['scale', 'auto']
}

grid = GridSearchCV(SVC(), param_grid, cv=5)
grid.fit(X_train, y_train)

print("Best Parameters:", grid.best_params_)
print("Best Accuracy:", grid.best_score_)
```
